```python
%%html
<!DOCTYPE html>
<html lang="pt-BR">
<head>
 <meta charset="UTF-8">
 <meta name="viewport" content="width=device-width, initial-scale=1.0">
 <title>Desafio Dinâmico: R$ 5.000 em 6 Meses</title>
 <!-- Importando biblioteca de confetes -->
 <script src="https://cdn.jsdelivr.net/npm/canvas-confetti@1.6.0/dist/confetti.browser.min.js"></script>
 <style>
 body {
 font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
 background-color: #f5f0fa;
 color: #333;
 margin: 0;
 padding: 20px;
 display: flex;
 flex-direction: column;
 align-items: center;
 }
 .header {
 text-align: center;
 margin-bottom: 25px;
 }
 h1 {
 color: #8e24aa;
 margin-bottom: 5px;
 }
 .dashboard {
 background: linear-gradient(135deg, #a18cd1, #fbc2eb);
 color: white;
 border-radius: 15px;
 padding: 20px 40px;
 text-align: center;
 box-shadow: 0 4px 15px rgba(0,0,0,0.1);
 margin-bottom: 30px;
 width: 100%;
 max-width: 500px;
 }
 .dashboard .total-box {
 font-size: 2.5rem;
 font-weight: bold;
 margin: 10px 0;
 }
 .progress-container {
 background-color: rgba(255,255,255,0.4);
 border-radius: 10px;
 height: 20px;
 width: 100%;
 margin-top: 10px;
 overflow: hidden;
 }
 .progress-bar {
 background-color: #fff;
 height: 100%;
 width: 0%;
 transition: width 0.3s ease;
 }
 .stats {
 display: flex;
 justify-content: space-between;
 margin-top: 10px;
 font-size: 0.9rem;
 }
 .grid {
 display: grid;
 grid-template-columns: repeat(auto-fill, minmax(70px, 1fr));
 gap: 10px;
 width: 100%;
 max-width: 900px;
 background: white;
 padding: 20px;
 border-radius: 15px;
 box-shadow: 0 4px 10px rgba(0,0,0,0.05);
 }
 .day-card {
 background-color: #f3e5f5;
 border: 2px solid #e1bee7;
 border-radius: 8px;
 padding: 10px 5px;
 text-align: center;
 cursor: pointer;
 user-select: none;
 transition: all 0.2s ease;
 }
 .day-card:hover {
 transform: scale(1.05);
 border-color: #ec407a;
 }
 .day-card.completed {
 background-color: #ec407a !important;
 border-color: #d81b60 !important;
 color: white;
 }
 .day-num {
 font-size: 0.75rem;
 color: #7b1fa2;
 display: block;
 margin-bottom: 4px;
 }
 .day-card.completed .day-num {
 color: #fbe9e7;
 }
 .day-val {
 font-weight: bold;
 font-size: 1rem;
 }
 .btn-reset {
 margin-top: 20px;
 background-color: #ab47bc;
 color: white;
 border: none;
 padding: 10px 20px;
 border-radius: 5px;
 cursor: pointer;
 font-weight: bold;
 }
 .btn-reset:hover {
 background-color: #8e24aa;
 }
 </style>
</head>
<body>

 <div class="header">
 <h1>Desafio Poupança Diária</h1>
 <p>Junte R$ 5.000 em 6 meses (180 dias) com valores variados</p>
 </div>

 <div class="dashboard">
 <div>Montante na Caixinha</div>
 <div class="total-box" id="totalBox">R$ 0,00</div>
 <div class="progress-container">
 <div class="progress-bar" id="progressBar"></div>
 </div>
 <div class="stats">
 <span id="daysCount">0 / 180 dias</span>
 <span>Meta: R$ 5.000,00</span>
 </div>
 </div>

 <div class="grid" id="gridContainer">
 <!-- Days generated by JS -->
 </div>

 <button class="btn-reset" onclick="resetChallenge()">Reiniciar Desafio</button>

 <script>
 const values = [6, 32, 22, 42, 37, 27, 7, 41, 22, 42, 17, 37, 7, 32, 21, 42, 22, 27, 7, 42, 22, 41, 27, 37, 7, 32, 22, 42, 31, 27, 7, 42, 22, 42, 37, 36, 7, 32, 22, 42, 17, 27, 6, 42, 22, 42, 22, 37, 7, 31, 22, 42, 27, 27, 7, 42, 21, 42, 32, 37, 7, 32, 22, 41, 37, 27, 7, 42, 22, 42, 17, 37, 7, 32, 22, 42, 22, 27, 7, 42, 22, 42, 27, 37, 7, 32, 22, 42, 32, 27, 7, 42, 22, 42, 37, 37, 7, 32, 22, 42, 17, 27, 7, 42, 22, 42, 22, 37, 7, 32, 22, 42, 27, 27, 7, 42, 22, 42, 32, 37, 7, 32, 22, 42, 37, 27, 7, 42, 22, 42, 17, 37, 7, 32, 22, 42, 22, 27, 7, 42, 22, 42, 27, 37, 7, 32, 22, 42, 32, 27, 7, 42, 22, 42, 37, 37, 7, 32, 22, 42, 17, 27, 7, 42, 22, 42, 22, 37, 7, 32, 22, 42, 27, 27, 7, 42, 22, 42, 32, 37];
 let completedDays = JSON.parse(localStorage.getItem('saved_5mil_days')) || [];

 const gridContainer = document.getElementById('gridContainer');
 const totalBox = document.getElementById('totalBox');
 const progressBar = document.getElementById('progressBar');
 const daysCount = document.getElementById('daysCount');

 function updateDashboard() {
 let totalSaved = 0;
 completedDays.forEach(idx => {
 totalSaved += values[idx];
 });

 totalBox.innerText = `R$ ${totalSaved.toLocaleString('pt-BR', {minimumFractionDigits: 2, maximumFractionDigits: 2})}`;

 const progressPercent = Math.min((totalSaved / 5000) * 100, 100);
 progressBar.style.width = `${progressPercent}%`;

 daysCount.innerText = `${completedDays.length} / 180 dias`;
 }

 function toggleDay(idx, element) {
 const position = completedDays.indexOf(idx);
 if (position === -1) {
 completedDays.push(idx);
 element.classList.add('completed');

 // Dispara os confetes ao completar um dia!
 confetti({
 particleCount: 100,
 spread: 70,
 origin: { y: 0.8 },
 colors: ['#ec407a', '#ab47bc', '#e1bee7', '#fbc2eb']
 });
 } else {
 completedDays.splice(position, 1);
 element.classList.remove('completed');
 }
 localStorage.setItem('saved_5mil_days', JSON.stringify(completedDays));
 updateDashboard();
 }

 function initGrid() {
 gridContainer.innerHTML = '';
 values.forEach((val, idx) => {
 const card = document.createElement('div');
 card.className = 'day-card';
 if (completedDays.includes(idx)) {
 card.classList.add('completed');
 }
 card.onclick = () => toggleDay(idx, card);

 card.innerHTML = `
 <span class="day-num">Dia ${idx + 1}</span>
 <span class="day-val">R$ ${val}</span>
 `;
 gridContainer.appendChild(card);
 });
 updateDashboard();
 }

 function resetChallenge() {
 if (confirm('Tem certeza que deseja zerar o seu progresso?')) {
 completedDays = [];
 localStorage.removeItem('saved_5mil_days');
 initGrid();
 }
 }

 initGrid();
 </script>
</body>
</html>
```
